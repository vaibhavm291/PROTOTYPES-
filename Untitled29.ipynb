```python
# Install required libraries at the very beginning to ensure they are available for import
# WBT binary ko executable banayein
!chmod +x /usr/local/lib/python3.12/dist-packages/whitebox/whitebox_tools
!pip install ultralytics
!pip install whitebox

import logging
import rasterio
import numpy as np
import geopandas as gpd
import os
from google.colab import files
from rasterio.mask import mask
from rasterio.warp import transform_geom, calculate_default_transform, reproject, Resampling
from shapely.geometry import shape, mapping, box
from ultralytics import YOLO
# pyproj imports are moved into reproject_to_utm where they are directly used

# Import WhiteboxTools after installation
from whitebox.whitebox_tools import WhiteboxTools
wbt = WhiteboxTools()
wbt.set_verbose_mode(True) # Changed to True for debugging

# Professional Logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class UnifiedTerrainEngine:
    def __init__(self, dem_path, model_path=None):
        self.raw_dem_path = os.path.abspath(dem_path)
        self.model = YOLO(model_path) if model_path else None

        # 1. Sabse pehle Reproject karein taaki calculations METERS mein ho (GIS-Purist Way)
        logging.info("Auto-Reprojecting DEM to UTM for 100% Area Accuracy...")
        self.dem_path = self.reproject_to_utm()

        # ===== CRS DIAGNOSTIC AFTER REPROJECTION =====
        with rasterio.open(self.dem_path) as src:
            print("\n--- CRS DIAGNOSTIC AFTER REPROJECTION ---")
            print("CRS:", src.crs)
            print("Is Geographic:", src.crs.is_geographic)
            print("Resolution:", src.res)
            print("Transform:", src.transform)
            print("------------------------------------------\n")

        # 2. AB YEH 30 NAHI HAI: Metadata se derive karein (Senior's Requirement)
        with rasterio.open(self.dem_path) as src:
            self.dem_crs = src.crs
            self.dem_res = src.res[0] # Auto-detected (10m, 30m, or 90m)
            self.dem_nodata = src.nodata
            self.dem_transform = src.transform

        logging.info(f"Engine Ready. Detected Resolution: {self.dem_res:.2f} meters.")

    def reproject_to_utm(self, output_path="projected_dem.tif"):
        """Automatically detects the correct UTM zone and reprojects."""
        from pyproj import CRS
        from pyproj.aoi import AreaOfInterest
        from pyproj.database import query_utm_crs_info

        with rasterio.open(self.raw_dem_path) as src:
            lon, lat = src.lnglat()
            utm_crs_list = query_utm_crs_info(
                datum_name="WGS 84",
                area_of_interest=AreaOfInterest(lon, lat, lon, lat)
            )
            dst_crs = CRS.from_user_input(f"{utm_crs_list[0].auth_name}:{utm_crs_list[0].code}")

            transform, width, height = calculate_default_transform(
                src.crs, dst_crs, src.width, src.height, *src.bounds)

            kwargs = src.meta.copy()
            kwargs.update({
                'crs': dst_crs,
                'transform': transform,
                'width': width,
                'height': height
            })

            with rasterio.open(output_path, 'w', **kwargs) as dst:
                for i in range(1, src.count + 1):
                    reproject(
                        source=rasterio.band(src, i), destination=rasterio.band(dst, i),
                        src_transform=src.transform, src_crs=src.crs,
                        dst_transform=transform, dst_crs=dst_crs, resampling=Resampling.bilinear)
        return os.path.abspath(output_path) # Ensure returned path is absolute

    def _spatial_handshake(self, image_path, detections):
        """Pixels to Global Coordinates conversion."""
        with rasterio.open(image_path) as img_src:
            img_crs, img_transform = img_src.crs, img_src.transform

        polygons = []
        for box_data in detections.boxes.xyxy.cpu().numpy():
            x1, y1, x2, y2 = box_data
            lon1, lat1 = rasterio.transform.xy(img_transform, y1, x1)
            lon2, lat2 = rasterio.transform.xy(img_transform, y2, x2)
            poly = box(lon1, lat2, lon2, lat1)

            # Aligns CV detection with DEM CRS
            geom_repro = transform_geom(img_crs, self.dem_crs, mapping(poly))
            polygons.append(shape(geom_repro))

        return gpd.GeoDataFrame(geometry=polygons, crs=self.dem_crs)

    def run_scientific_hydrology(self):
        """Standard Scientific Flow Routing."""
        base_dir = os.path.abspath(os.getcwd()) # Direct /content/ directory
        filled_dem_path = os.path.join(base_dir, "filled_dem.tif")
        flow_acc_path = os.path.join(base_dir, "flow_accum.tif")
        slope_path = os.path.join(base_dir, "slope.tif")

        if not os.path.exists(self.dem_path):
            raise FileNotFoundError(f"Input DEM not found at {self.dem_path}")

        logging.info("Starting WBT Processing...")

        # 1. Fill Depressions
        wbt.fill_depressions(self.dem_path, filled_dem_path)

        # Agar abhi bhi error aaye, toh iska matlab WBT ne file banayi hi nahi
        if not os.path.exists(filled_dem_path):
            # Force check for any WBT errors
            print("WBT Work Dir:", wbt.work_dir)
            raise FileNotFoundError(f"CRITICAL: WBT failed. Check if {self.dem_path} is a valid GeoTIFF.")

        # 2. Real D8 Flow Accumulation
        wbt.d8_flow_accumulation(filled_dem_path, flow_acc_path)
        if not os.path.exists(flow_acc_path):
            raise FileNotFoundError(f"WhiteboxTools failed to create {flow_acc_path}")

        # 3. Scientific Slope
        wbt.slope(self.dem_path, slope_path)
        if not os.path.exists(slope_path):
            raise FileNotFoundError(f"WhiteboxTools failed to create {slope_path}")

        return slope_path, flow_acc_path

    def generate_report(self, mode_choice, image_path=None, civic_infra=None):
        slope_tif, flow_tif = self.run_scientific_hydrology()

        with rasterio.open(slope_tif) as s_src, rasterio.open(flow_tif) as f_src:
            slope_map = s_src.read(1)
            flow_map = f_src.read(1)

        # Calculate global pixel_area and actual_area_m2 for the entire DEM
        # This is needed for global feasibility scoring BEFORE iterating through geometries
        with rasterio.open(self.dem_path) as src:
            global_res_x, global_res_y = src.res
            global_pixel_area = float(global_res_x * global_res_y)
            global_actual_area_m2 = float(src.width * src.height * global_pixel_area)


        # --- STEP 1: SLOPE ZONING ---
        # Valid pixels mein se check karein kitne steep hain
        steep_pixels = slope_map[slope_map > 15]
        percent_steep = (steep_pixels.size / slope_map.size) * 100

        # Buildable Zone: Jahan slope < 10 degree ho aur pooling risk na ho
        buildable_mask = (slope_map < 10) & (flow_map < np.percentile(flow_map, 90))
        percent_buildable = (np.count_nonzero(buildable_mask) / slope_map.size) * 100

        # Cut-Fill Classification: Jahan slope > 15 hai wahan 'High Cut-Fill' required
        cut_fill_type = "High (Terracing Required)" if percent_steep > 20 else "Standard"

        # --- STEP 2 & 3: DRAINAGE CORRIDORS ---
        # Jaha Flow Accumulation top 5% mein hai, wo natural 'Runoff Corridors' hain
        threshold = np.percentile(flow_map, 95)
        hotspot_mask = flow_map > threshold

        # Waterlogging Hotspot: Jahan slope 0-1 degree hai AUR accumulation high hai
        waterlogging_mask = (slope_map < 1.5) & (flow_map > threshold)
        risk_area_m2 = np.count_nonzero(waterlogging_mask) * global_pixel_area # Use global_pixel_area

        # --- STEP 5 & 6: SCORING MODEL ---
        # Weighted Formula: High Slope aur High Accumulation score ko ghatate hain
        # Area jahan construction mushkil hai (Risk Factor)
        risk_factor = (percent_steep * 0.4) + ( (risk_area_m2 / global_actual_area_m2) * 100 * 0.6 ) # Use global_actual_area_m2

        # Final Feasibility Score (Out of 100)
        feasibility_score = max(0, 100 - risk_factor)

        # Flood Susceptibility Index (1-5)
        flood_index = 5 if risk_area_m2 > (0.2 * global_actual_area_m2) else 1

        # Define these global summary texts here, using the global calculations
        suitability_text = f"{percent_buildable:.1f}% of land is suitable for direct foundation placement."
        convergence_text = "Water convergence risk minimal." if flood_index < 3 else "High runoff convergence detected - drainage priority."
        tolerance_text = f"Cut-fill requirement: {cut_fill_type} within engineering tolerance."


        # Mode Selection Logic
        if mode_choice == "1" and image_path:
            results = self.model(image_path)[0]
            gdf = self._spatial_handshake(image_path, results)
        else:
            logging.info("Running Mode 2: DEM + GIS Only")
            with rasterio.open(self.dem_path) as src:
                gdf = gpd.GeoDataFrame(geometry=[box(*src.bounds)], crs=self.dem_crs)


        # --- SCIENTIFIC METRIC CALCULATION (WITH Z-AXIS VALIDATION & MASKING) ---
        report_list = []
        with rasterio.open(self.dem_path) as src:
            res_x, res_y = src.res
            pixel_area = float(res_x * res_y)

            for geom in gdf.geometry:
                out_img, _ = mask(src, [mapping(geom)], crop=True)
                z_raw = out_img[0]

                # Nodata aur NaNs ko filter karke asli ground pixels nikalna
                valid_mask = (z_raw != self.dem_nodata) & (~np.isnan(z_raw)) & (z_raw > -1000)
                z_valid = z_raw[valid_mask]

                if z_valid.size > 0:
                    # --- [POINT 4: Z-BUFFER MASKING REFINEMENT] ---
                    # Ground level ko 5th percentile se subtract karke relative height nikalna
                    ground_level = np.percentile(z_valid, 5)
                    z_relative = z_valid - ground_level

                    z_min, z_max = np.nanmin(z_valid), np.nanmax(z_valid)
                    h_diff = float(z_max - z_min)

                    # --- [POINT 2: Z-AXIS VALIDATION] ---
                    # Agar height 0.3m se zyada hai tabhi saccha 3D object hai
                    is_3d_object = h_diff > 0.3
                    status_obj = "Verified 3D Object" if is_3d_object else "Flat Surface/Noise" # Renamed to avoid clash with global 'status'
                    # Extreme values ko ignore karke sahi mean slope nikalna
                    valid_slopes = slope_map[(slope_map > 0) & (slope_map < 90)]
                    mean_slope = float(np.nanmean(valid_slopes)) if valid_slopes.size > 0 else 0.0

                    # True Volume using Z-Buffer logic
                    actual_area_m2 = float(z_valid.size * pixel_area)
                    true_vol_m3 = float(np.sum(z_relative[z_relative > 0]) * pixel_area)

                    # Updated Recommendation Logic (for individual geometries)
                    if is_3d_object and h_diff > 2.0:
                        rec = "Significant Infrastructure Detected - Check Stability"
                    elif mean_slope < 1.5:
                        rec = "High Pooling Risk (Flat Area)"
                    else:
                        rec = "Optimal Terrain for Construction"

                    report_list.append({
                        "geometry": geom,
                        "Elev_Min": float(z_min),
                        "Elev_Max": float(z_max),
                        "Height_Diff": h_diff,
                        "Area_m2": actual_area_m2,
                        "Vol_Est_m3": true_vol_m3,
                        "Max_Flow_Acc": float(np.nanmax(flow_map)),
                        "Mean_Slope_Deg": mean_slope,
                        "Object_Status": status_obj, # Use renamed variable
                        "Engineering_Rec": rec, # Object-level recommendation
                        # Client-facing metrics (global calculations)
                        "Feasibility_Score": round(feasibility_score, 2),
                        "Percent_Steep_Land": round(percent_steep, 2),
                        "Buildable_Area_Percent": round(percent_buildable, 2),
                        "Flood_Susceptibility": flood_index,
                        "Waterlogging_Risk_m2": risk_area_m2,
                        "Cut_Fill_Category": cut_fill_type,
                        "Suitability_Statement": suitability_text,
                        "Hydrology_Summary": convergence_text,
                        "Slope_Verdict": tolerance_text
                    })

        final_gdf = gpd.GeoDataFrame(report_list, crs=self.dem_crs)

        # --- MANDATORY SPATIAL JOIN ---
        if civic_infra is not None:
            logging.info("Performing Industrial Spatial Join with Civic Infrastructure...")
            infra_gdf = civic_infra.to_crs(self.dem_crs)
            final_gdf = gpd.sjoin(final_gdf, infra_gdf, how="inner", predicate="intersects")


        return final_output_handling(final_gdf, feasibility_score, suitability_text, convergence_text, tolerance_text)

def final_output_handling(gdf, feasibility_score, suitability_text, convergence_text, tolerance_text):
    if not gdf.empty:
        gdf.to_file("ENGINEER_VERIFICATION_REPORT.geojson", driver="GeoJSON")
        gdf.drop(columns='geometry').to_csv("SITE_METRICS.csv", index=False)
        return {
            "status": "✅ Output Generated: GeoJSON for Architect, CSV for Budgeting.",
            "feasibility_score": feasibility_score,
            "suitability_text": suitability_text,
            "convergence_text": convergence_text,
            "tolerance_text": tolerance_text
        }
    return {
        "status": "⚠ No matching features found.",
        "feasibility_score": 0.0, # Default for no features found
        "suitability_text": "N/A",
        "convergence_text": "N/A",
        "tolerance_text": "N/A"
    }

# ==========================================================S
# USER INTERFACE
# ==========================================================
print("--- TERRAIN INTELLIGENCE SYSTEM ---\n")
dem_file = list(files.upload().keys())[0]

print("Select Mode: \n1. CV + DEM + GIS (Urban/Drone)\n2. DEM + GIS Only (Open Field)")
choice = input("Enter choice (1/2): ")

engine = None
if choice == "1":
    img_file = list(files.upload().keys())[0]
    model_file = list(files.upload().keys())[0] # Your .pt file
    engine = UnifiedTerrainEngine(dem_file, model_file)
    status_dict = engine.generate_report("1", image_path=img_file)
else:
    engine = UnifiedTerrainEngine(dem_file)
    status_dict = engine.generate_report("2")

status = status_dict["status"]
feasibility_score = status_dict["feasibility_score"]
suitability_text = status_dict["suitability_text"]
convergence_text = status_dict["convergence_text"]
tolerance_text = status_dict["tolerance_text"]

print(status)
import pandas as pd
from google.colab import data_table
import matplotlib.pyplot as plt

# CSV file ko screen par load karke dikhane ke liye
try:
    df = pd.read_csv("SITE_METRICS.csv")
    print("\n--- ☢️ INDUSTRIAL SITE SUMMARY ---")
    display(data_table.DataTable(df, include_index=False, num_rows_per_page=10)) # Using data_table for display
    # --- EXECUTIVE SUMMARY DATA ---
    print(f"\n{'='*50}")
    print(f"☢️  EXECUTIVE SITE FEASIBILITY REPORT")
    print(f"{'='*50}")
    print(f"️  Analysis Status: {status}")
    print(f"️  Overall Score: {feasibility_score:.1f}/100") # Use global var
    print(f"✅ Summary: {suitability_text}") # Use global var
    print(f"️  Hydrology: {convergence_text}") # Use global var
    print(f"️  Terrain: {tolerance_text}") # Use global var
    print(f"\n{'*'*50}")
    print(f"⚖️  LEGAL DISCLAIMER & LIMITATIONS")
    print(f"1. This report is based on Digital Elevation Models (DEM) and Satellite data.")
    print(f"2. Results must be verified by on-site geotechnical bore-log testing.")
    print(f"3. Calculations are for preliminary planning and bidding purposes only.")
    print(f"4. The system is not liable for structural failures based on this simulation.")
    print(f"{'*'*50}")
    print(f"\n✅ Total Site Area Analyzed: {df['Area_m2'].sum():,.2f} m²")
    print(f"✅ Maximum Height Difference: {df['Height_Diff'].max():.2f} meters")
except FileNotFoundError:
    print("❌ CSV not found. Ensure the engine ran successfully.")
except Exception as e:
    print(f"⚠ An error occurred while processing CSV: {e}")

import geopandas as gpd

try:
    gdf = gpd.read_file("ENGINEER_VERIFICATION_REPORT.geojson")

    # Create a professional plot
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    gdf.plot(column='Mean_Slope_Deg', ax=ax, legend=True, # Corrected column name
             cmap='YlOrRd', legend_kwds={'label': "Slope Gradient (Degrees)"})

    plt.title("Spatial Intelligence Report: Terrain Risk Map", fontsize=15)
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.grid(alpha=0.3)
    plt.show()

    import matplotlib.pyplot as plt

    def generate_professional_maps(gdf):
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))

        # MAP 1: Buildable Zone Map (Decision for Architect)
        gdf.plot(column='Feasibility_Score', ax=ax1, cmap='RdYlGn', legend=True,
                 legend_kwds={'label': "Feasibility Score (0-100)"})
        ax1.set_title("Buildable Zone Intelligence Map", fontsize=14)

        # MAP 2: Terrain Risk (Slope + Hydrology)
        gdf.plot(column='Mean_Slope_Deg', ax=ax2, cmap='YlOrBr', legend=True,
                 legend_kwds={'label': "Slope Gradient (Degrees)"})
        ax2.set_title("Hydrological & Slope Risk Map", fontsize=14)

        plt.tight_layout()
        plt.savefig("PILOT_MAPS_REPORT.png")
        plt.show()

    # Is function ko print(status) ke baad call karein
    generate_professional_maps(gdf)

    print(f"✅ Map verified. Total features analyzed: {len(gdf)}")
except FileNotFoundError:
    print("❌ GeoJSON not found. Ensure the engine ran successfully.")
except Exception as e:
    print(f"⚠ Visualization failed: {e}")
```

```python
# Install required libraries at the very beginning to ensure they are available for import
# WBT binary ko executable banayein
!chmod +x /usr/local/lib/python3.12/dist-packages/whitebox/whitebox_tools
!rm projected_dem.tif slope.tif flow_accum.tif filled_dem.tif
!pip install ultralytics
!pip install whitebox

import logging
import rasterio
import numpy as np
import geopandas as gpd
import os
from google.colab import files
from rasterio.mask import mask

from rasterio.warp import transform_geom, calculate_default_transform, reproject, Resampling
from shapely.geometry import shape, mapping, box
from ultralytics import YOLO
from pyproj import CRS
from pyproj.aoi import AreaOfInterest
from pyproj.database import query_utm_crs_info
import pandas as pd
from google.colab import data_table
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import rasterio.plot # Added this import

# Import WhiteboxTools after installation
from whitebox.whitebox_tools import WhiteboxTools
wbt = WhiteboxTools()
wbt.set_verbose_mode(True) # Changed to True for debugging
wbt.set_working_dir(os.getcwd()) # Added this line

# Professional Logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')

class UnifiedTerrainEngine:
    def __init__(self, dem_path, model_path=None):
        self.raw_dem_path = os.path.abspath(dem_path)
        self.model = YOLO(model_path) if model_path else None

        # 1. Sabse pehle Reproject karein taaki calculations METERS mein ho (GIS-Purist Way)
        logging.info("Auto-Reprojecting DEM to UTM for 100% Area Accuracy...")
        self.dem_path = self.reproject_to_utm()

        # ===== CRS DIAGNOSTIC AFTER REPROJECTION =====
        with rasterio.open(self.dem_path) as src:
            print("\n--- CRS DIAGNOSTIC AFTER REPROJECTION ---")
            print("CRS:", src.crs)
            print("Is Geographic:", src.crs.is_geographic)
            print("Resolution:", src.res)
            print("Transform:", src.transform)
            print("------------------------------------------\n")

        # 2. AB YEH 30 NAHI HAI: Metadata se derive karein (Senior's Requirement)
        with rasterio.open(self.dem_path) as src:
            self.dem_crs = src.crs
            self.dem_res = src.res[0] # Auto-detected (10m, 30m, or 90m)
            self.dem_nodata = src.nodata
            self.dem_transform = src.transform

        logging.info(f"Engine Ready. Detected Resolution: {self.dem_res:.2f} meters.")

    def reproject_to_utm(self, output_path="projected_dem.tif"):
        """Automatically detects the correct UTM zone and reprojects."""
        with rasterio.open(self.raw_dem_path) as src:
            lon, lat = src.lnglat()
            utm_crs_list = query_utm_crs_info(
                datum_name="WGS 84",
                area_of_interest=AreaOfInterest(lon, lat, lon, lat)
            )
            dst_crs = CRS.from_user_input(f"{utm_crs_list[0].auth_name}:{utm_crs_list[0].code}")

            transform, width, height = calculate_default_transform(
                src.crs, dst_crs, src.width, src.height, *src.bounds)

            kwargs = src.meta.copy()
            kwargs.update({
                'crs': dst_crs,
                'transform': transform,
                'width': width,
                'height': height
            })

            with rasterio.open(output_path, 'w', **kwargs) as dst:
                for i in range(1, src.count + 1):
                    reproject(
                        source=rasterio.band(src, i), destination=rasterio.band(dst, i),
                        src_transform=src.transform, src_crs=src.crs,
                        dst_transform=transform, dst_crs=dst_crs, resampling=Resampling.bilinear)
        return output_path

    def _spatial_handshake(self, image_path, detections):
        """Pixels to Global Coordinates conversion."""
        with rasterio.open(image_path) as img_src:
            img_crs, img_transform = img_src.crs, img_src.transform

        polygons = []
        for box_data in detections.boxes.xyxy.cpu().numpy():
            x1, y1, x2, y2 = box_data
            lon1, lat1 = rasterio.transform.xy(img_transform, y1, x1)
            lon2, lat2 = rasterio.transform.xy(img_transform, y2, x2)
            poly = box(lon1, lat2, lon2, lat1)

            # Aligns CV detection with DEM CRS
            geom_repro = transform_geom(img_crs, self.dem_crs, mapping(poly))
            polygons.append(shape(geom_repro))

        return gpd.GeoDataFrame(geometry=polygons, crs=self.dem_crs)

    def run_scientific_hydrology(self):
        """Standard Scientific Flow Routing."""
        # Define output paths in the current working directory
        filled_dem_path = os.path.join(os.getcwd(), "filled_dem.tif")
        flow_acc_path = os.path.join(os.getcwd(), "flow_accum.tif")
        slope_path = os.path.join(os.getcwd(), "slope.tif")

        logging.info(f"Input DEM for WBT: {self.dem_path}")
        logging.info(f"Output filled_dem_path: {filled_dem_path}")
        logging.info(f"Output flow_acc_path: {flow_acc_path}")
        logging.info(f"Output slope_path: {slope_path}")

        # 1. Fill Depressions
        wbt.fill_depressions(self.dem_path, filled_dem_path)
        if not os.path.exists(filled_dem_path):
            raise FileNotFoundError(f"WhiteboxTools failed to create {filled_dem_path}")

        # 2. Real D8 Flow Accumulation
        wbt.d8_flow_accumulation(filled_dem_path, flow_acc_path)
        if not os.path.exists(flow_acc_path):
            raise FileNotFoundError(f"WhiteboxTools failed to create {flow_acc_path}")

        # 3. Scientific Slope
        wbt.slope(self.dem_path, slope_path)
        if not os.path.exists(slope_path):
            raise FileNotFoundError(f"WhiteboxTools failed to create {slope_path}")

        return slope_path, flow_acc_path

    def generate_report(self, mode_choice, image_path=None, civic_infra=None):
        slope_tif, flow_tif = self.run_scientific_hydrology()

        # Open s_src and f_src outside the loop so they don't close prematurely
        s_src = rasterio.open(slope_tif)
        f_src = rasterio.open(flow_tif)

        try:
            slope_map = s_src.read(1)
            flow_map = f_src.read(1)

            # Calculate global pixel_area and actual_area_m2 for the entire DEM
            # This is needed for global feasibility scoring BEFORE iterating through geometries
            with rasterio.open(self.dem_path) as src:
                global_res_x, global_res_y = src.res
                global_pixel_area = float(global_res_x * global_res_y)
                global_actual_area_m2 = float(src.width * src.height * global_pixel_area)


            # --- STEP 1: SLOPE ZONING ---
            # Valid pixels mein se check karein kitne steep hain
            steep_pixels = slope_map[slope_map > 15]
            percent_steep = (steep_pixels.size / slope_map.size) * 100

            # Buildable Zone: Jahan slope < 10 degree ho aur pooling risk na ho
            buildable_mask = (slope_map < 10) & (flow_map < np.percentile(flow_map, 90))
            percent_buildable = (np.count_nonzero(buildable_mask) / slope_map.size) * 100

            # Cut-Fill Classification: Jahan slope > 15 hai wahan 'High Cut-Fill' required
            cut_fill_type = "High (Terracing Required)" if percent_steep > 20 else "Standard"

            # --- STEP 2 & 3: DRAINAGE CORRIDORS ---
            # Jaha Flow Accumulation top 5% mein hai, wo natural 'Runoff Corridors' hain
            threshold = np.percentile(flow_map, 95)
            hotspot_mask = flow_map > threshold

            # Waterlogging Hotspot: Jahan slope 0-1 degree hai AUR accumulation high hai
            waterlogging_mask = (slope_map < 1.5) & (flow_map > threshold)
            risk_area_m2 = np.count_nonzero(waterlogging_mask) * global_pixel_area # Use global_pixel_area

            # --- STEP 5 & 6: SCORING MODEL ---
            # Weighted Formula: High Slope aur High Accumulation score ko ghatate hain
            # Area jahan construction mushkil hai (Risk Factor)
            risk_factor = (percent_steep * 0.4) + ( (risk_area_m2 / global_actual_area_m2) * 100 * 0.6 ) # Use global_actual_area_m2

            # Final Feasibility Score (Out of 100)
            feasibility_score = max(0, 100 - risk_factor)

            # Flood Susceptibility Index (1-5)
            flood_index = 5 if risk_area_m2 > (0.2 * global_actual_area_m2) else 1

            # Define these global summary texts here, using the global calculations
            suitability_text = f"{percent_buildable:.1f}% of land is suitable for direct foundation placement."
            convergence_text = "Water convergence risk minimal." if flood_index < 3 else "High runoff convergence detected - drainage priority."
            tolerance_text = f"Cut-fill requirement: {cut_fill_type} within engineering tolerance."


            # Mode Selection Logic
            if mode_choice == "1" and image_path:
                results = self.model(image_path)[0]
                gdf = self._spatial_handshake(image_path, results)
            else:
                logging.info("Running Mode 2: DEM + GIS Only")
                with rasterio.open(self.dem_path) as src:
                    gdf = gpd.GeoDataFrame(geometry=[box(*src.bounds)], crs=self.dem_crs)


            # --- SCIENTIFIC METRIC CALCULATION (WITH Z-AXIS VALIDATION & MASKING) ---
            report_list = []
            with rasterio.open(self.dem_path) as src:
                res_x, res_y = src.res
                pixel_area = float(res_x * res_y)

                for geom in gdf.geometry:
                    out_img, _ = mask(src, [mapping(geom)], crop=True)
                    z_raw = out_img[0]

                    # Nodata aur NaNs ko filter karke asli ground pixels nikalna
                    valid_mask = (z_raw != self.dem_nodata) & (~np.isnan(z_raw)) & (z_raw > -1000)
                    z_valid = z_raw[valid_mask]

                    if z_valid.size > 0:
                        # --- [POINT 4: Z-BUFFER MASKING REFINEMENT] ---
                        # Ground level ko 5th percentile se subtract karke relative height nikalna
                        ground_level = np.percentile(z_valid, 5)
                        z_relative = z_valid - ground_level

                        z_min, z_max = np.nanmin(z_valid), np.nanmax(z_valid)
                        h_diff = float(z_max - z_min)

                        # --- [POINT 2: Z-AXIS VALIDATION] ---
                        # Agar height 0.3m se zyada hai tabhi saccha 3D object hai
                        is_3d_object = h_diff > 0.3
                        status_obj = "Verified 3D Object" if is_3d_object else "Flat Surface/Noise" # Renamed to avoid clash with global 'status'

                        # True Volume using Z-Buffer logic
                        actual_area_m2 = float(z_valid.size * pixel_area)
                        true_vol_m3 = float(np.sum(z_relative[z_relative > 0]) * pixel_area)

                        # POORE map ke bajaye sirf us GEOMETRY ka slope nikalna
                        out_slope, _ = mask(s_src, [mapping(geom)], crop=True)
                        z_slope = out_slope[0]
                        valid_z_slope = z_slope[(z_slope > 0) & (z_slope < 90)]
                        mean_slope = float(np.nanmean(valid_z_slope)) if valid_z_slope.size > 0 else 0.0

                        # Updated Recommendation Logic (for individual geometries)
                        if is_3d_object and h_diff > 2.0:
                            rec = "Significant Infrastructure Detected - Check Stability"
                        elif mean_slope < 1.5:
                            rec = "High Pooling Risk (Flat Area)"
                        else:
                            rec = "Optimal Terrain for Construction"

                        report_list.append({
                            "geometry": geom,
                            "Elev_Min": float(z_min),
                            "Elev_Max": float(z_max),
                            "Height_Diff": h_diff,
                            "Area_m2": actual_area_m2,
                            "Vol_Est_m3": true_vol_m3,
                            "Max_Flow_Acc": float(np.nanmax(flow_map)),
                            "Mean_Slope_Deg": mean_slope,
                            "Object_Status": status_obj, # Use renamed variable
                            "Engineering_Rec": rec, # Object-level recommendation
                            # Client-facing metrics (global calculations)
                            "Feasibility_Score": round(feasibility_score, 2),
                            "Percent_Steep_Land": round(percent_steep, 2),
                            "Buildable_Area_Percent": round(percent_buildable, 2),
                            "Flood_Susceptibility": flood_index,
                            "Waterlogging_Risk_m2": risk_area_m2,
                            "Cut_Fill_Category": cut_fill_type,
                            "Suitability_Statement": suitability_text,
                            "Hydrology_Summary": convergence_text,
                            "Slope_Verdict": tolerance_text
                        })

            final_gdf = gpd.GeoDataFrame(report_list, crs=self.dem_crs)

            # --- MANDATORY SPATIAL JOIN ---
            if civic_infra is not None:
                logging.info("Performing Industrial Spatial Join with Civic Infrastructure...")
                infra_gdf = civic_infra.to_crs(self.dem_crs)
                final_gdf = gpd.sjoin(final_gdf, infra_gdf, how="inner", predicate="intersects")


            return final_output_handling(final_gdf, feasibility_score, suitability_text, convergence_text, tolerance_text)
        finally:
            s_src.close()
            f_src.close()

def final_output_handling(gdf, feasibility_score, suitability_text, convergence_text, tolerance_text):
    if not gdf.empty:
        gdf.to_file("ENGINEER_VERIFICATION_REPORT.geojson", driver="GeoJSON")
        gdf.drop(columns='geometry').to_csv("SITE_METRICS.csv", index=False)
        return {
            "status": "✅ Output Generated: GeoJSON for Architect, CSV for Budgeting.",
            "feasibility_score": feasibility_score,
            "suitability_text": suitability_text,
            "convergence_text": convergence_text,
            "tolerance_text": tolerance_text
        }
    return {
        "status": "⚠ No matching features found.",
        "feasibility_score": 0.0, # Default for no features found
        "suitability_text": "N/A",
        "convergence_text": "N/A",
        "tolerance_text": "N/A"
    }

# ==========================================================S
# USER INTERFACE
# ==========================================================
print("--- TERRAIN INTELLIGENCE SYSTEM ---\n")
dem_file = list(files.upload().keys())[0]

print("Select Mode: \n1. CV + DEM + GIS (Urban/Drone)\n2. DEM + GIS Only (Open Field)")
choice = input("Enter choice (1/2): ")

engine = None
if choice == "1":
    img_file = list(files.upload().keys())[0]
    model_file = list(files.upload().keys())[0] # Your .pt file
    status_dict = engine.generate_report("1", image_path=img_file)
else:
    engine = UnifiedTerrainEngine(dem_file)
    status_dict = engine.generate_report("2")

status = status_dict["status"]
feasibility_score = status_dict["feasibility_score"]
suitability_text = status_dict["suitability_text"]
convergence_text = status_dict["convergence_text"]
tolerance_text = status_dict["tolerance_text"]

print(status)

# CSV file ko screen par load karke dikhane ke liye
try:
    df = pd.read_csv("SITE_METRICS.csv")
    print("\n--- 🏗️ INDUSTRIAL SITE SUMMARY ---")
    display(data_table.DataTable(df, include_index=False, num_rows_per_page=10)) # Using data_table for display
    # --- EXECUTIVE SUMMARY DATA ---
    print(f"\n{'='*50}")
    print(f"🏗️  EXECUTIVE SITE FEASIBILITY REPORT")
    print(f"{'='*50}")
    print(f"📍 Analysis Status: {status}")
    print(f"📊 Overall Score: {feasibility_score:.1f}/100") # Use global var
    print(f"✅ Summary: {suitability_text}") # Use global var
    print(f"🌊 Hydrology: {convergence_text}") # Use global var
    print(f"⛰️  Terrain: {tolerance_text}") # Use global var
    print(f"\n{'*'*50}")
    print(f"⚖️  LEGAL DISCLAIMER & LIMITATIONS")
    print(f"1. This report is based on Digital Elevation Models (DEM) and Satellite data.")
    print(f"2. Results must be verified by on-site geotechnical bore-log testing.")
    print(f"3. Calculations are for preliminary planning and bidding purposes only.")
    print(f"4. The system is not liable for structural failures based on this simulation.")
    print(f"{'*'*50}")
    print(f"\n✅ Total Site Area Analyzed: {df['Area_m2'].sum():,.2f} m²")
    print(f"✅ Maximum Height Difference: {df['Height_Diff'].max():.2f} meters")
except FileNotFoundError:
    print("❌ CSV not found. Ensure the engine ran successfully.")
except Exception as e:
    print(f"⚠ An error occurred while processing CSV: {e}")


try:
    gdf = gpd.read_file("ENGINEER_VERIFICATION_REPORT.geojson")

    # Create a professional plot
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    gdf.plot(column='Mean_Slope_Deg', ax=ax, legend=True, # Corrected column name
             cmap='YlOrRd', legend_kwds={'label': "Slope Gradient (Degrees)"})

    plt.title("Spatial Intelligence Report: Terrain Risk Map", fontsize=15)
    plt.xlabel("Longitude")
    plt.ylabel("Latitude")
    plt.grid(alpha=0.3)
    plt.show()

    def generate_professional_maps(gdf):
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 8))

        # MAP 1: Buildable Zone Map (Decision for Architect)
        gdf.plot(column='Feasibility_Score', ax=ax1, cmap='RdYlGn', legend=True,
                 legend_kwds={'label': "Feasibility Score (0-100)"})
        ax1.set_title("Buildable Zone Intelligence Map", fontsize=14)

        # MAP 2: Terrain Risk (Slope + Hydrology)
        gdf.plot(column='Mean_Slope_Deg', ax=ax2, cmap='YlOrBr', legend=True,
                 legend_kwds={'label': "Slope Gradient (Degrees)"})
        ax2.set_title("Hydrological & Slope Risk Map", fontsize=14)

        plt.tight_layout()
        plt.savefig("PILOT_MAPS_REPORT.png")
        plt.show()

    # Is function ko print(status) ke baad call karein
    generate_professional_maps(gdf)

    def generate_pixel_level_maps(engine_obj):
        # 1. Base data load karein
        slope_tif, flow_tif = engine_obj.run_scientific_hydrology()

        # Open s_src and f_src outside the inner `with` block
        s_src = rasterio.open(slope_tif)
        f_src = rasterio.open(flow_tif)

        try:
            slope_data = s_src.read(1)
            flow_data = f_src.read(1)
            extent = rasterio.plot.plotting_extent(s_src)

            # 2. Feasibility Map Array banana (Pixel-level Intelligence)
            # Wahi math jo humne loop mein lagaya tha, ab pure array par
            threshold_flow = np.percentile(flow_data, 95)
            risk_mask = (slope_data > 15).astype(float) * 0.4 + ((slope_data < 1.5) & (flow_data > threshold_flow)).astype(float) * 0.6
            feasibility_map = 100 - (risk_mask * 100)

            # 3. Plotting
            fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(22, 10))

            # Detailed Feasibility Map (Green = Good, Red = Risky)
            im1 = ax1.imshow(feasibility_map, extent=extent, cmap='RdYlGn', vmin=0, vmax=100)
            ax1.set_title("Detailed Buildable Zone Intelligence (Pixel-Level)", fontsize=15)
            plt.colorbar(im1, ax=ax1, label="Feasibility Score")

            # Detailed Slope Risk Map
            im2 = ax2.imshow(slope_data, extent=extent, cmap='YlOrBr', vmin=0, vmax=45)
            ax2.set_title("Terrain Slope & Hydrology Risk Detail", fontsize=15)
            plt.colorbar(im2, ax=ax2, label="Slope Degrees")

            plt.tight_layout()
            plt.savefig("DETAILED_PIXEL_REPORT.png")
            plt.show()
        finally:
            s_src.close()
            f_src.close()

    # Sabse aakhir mein ise call karein
    generate_pixel_level_maps(engine)


    print(f"✅ Map verified. Total features analyzed: {len(gdf)}")
except FileNotFoundError:
    print("❌ GeoJSON not found. Ensure the engine ran successfully.")
except Exception as e:
    print(f"❌ Visualization failed: {e}")
```
